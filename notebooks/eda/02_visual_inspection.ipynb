## Helper — single-panel map

```python
def plot_layer(da, title, cmap="viridis", vmin=None, vmax=None, save_as=None):
    """Quick single-panel map for a 2-D DataArray."""
    fig, ax = plt.subplots(figsize=(7, 6))
    da.plot(ax=ax, cmap=cmap, vmin=vmin, vmax=vmax, robust=True)
    ax.set_title(title)
    ax.set_aspect("equal")
    if save_as:
        fig.savefig(FIG_DIR / save_as, dpi=150, bbox_inches="tight")
    plt.show()
```
